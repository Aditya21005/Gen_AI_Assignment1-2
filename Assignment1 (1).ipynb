```python
from reportlab.lib.pagesizes import letter
from reportlab.platypus import SimpleDocTemplate, Paragraph
from reportlab.lib.styles import getSampleStyleSheet

doc = SimpleDocTemplate("data.pdf", pagesize=letter)
styles = getSampleStyleSheet()

text = """
Automata Theory and NLP Notes

DFA: Deterministic Finite Automaton is a model with fixed transitions.

NFA: Non-deterministic Finite Automaton allows multiple transitions.

NLP: Natural Language Processing enables machines to understand language.

Tokenization splits text into words.

Stemming reduces words to root form.

Lemmatization converts words to base form.
"""

doc.build([Paragraph(text, styles["Normal"])])
```

```python
from langchain_community.document_loaders import PyPDFLoader
from langchain.text_splitter import RecursiveCharacterTextSplitter

loader = PyPDFLoader("data.pdf")
documents = loader.load()

splitter1 = RecursiveCharacterTextSplitter(chunk_size=500, chunk_overlap=50)
chunks1 = splitter1.split_documents(documents)

splitter2 = RecursiveCharacterTextSplitter(chunk_size=1000, chunk_overlap=100)
chunks2 = splitter2.split_documents(documents)
```

```python
from langchain_community.embeddings import HuggingFaceEmbeddings
from langchain_community.vectorstores import FAISS, Chroma

emb1 = HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2")
emb2 = HuggingFaceEmbeddings(model_name="BAAI/bge-small-en")

db_faiss = FAISS.from_documents(chunks1, emb1)
db_chroma = Chroma.from_documents(chunks2, emb2)
```

```python
from transformers import pipeline

llm_gpt2 = pipeline("text-generation", model="gpt2")
llm_distil = pipeline("text-generation", model="distilgpt2")
llm_flan = pipeline("text-generation", model="google/flan-t5-base")
llm_falcon = pipeline("text-generation", model="tiiuae/falcon-rw-1b")
```

```python
def generate_answer(query, docs, llm):
    context = "\n".join([doc.page_content for doc in docs])

    prompt = f"""
    Answer clearly using the context below:

    {context}

    Question: {query}
    Answer:
    """

    output = llm(prompt, max_length=200)
    return output[0]['generated_text']
```

```python
queries = [
    "What is DFA?",
    "Explain NLP",
    "What is tokenization?"
]

for q in queries:
    print("\n" + "="*70)
    print("QUESTION:", q)

    docs_f = db_faiss.similarity_search(q, k=3)
    docs_c = db_chroma.similarity_search(q, k=3)

    print("\n🔹 GPT-2 (FAISS + MiniLM)")
    print(generate_answer(q, docs_f, llm_gpt2))

    print("\n🔹 DistilGPT2 (FAISS + MiniLM)")
    print(generate_answer(q, docs_f, llm_distil))

    print("\n🔹 FLAN-T5 (Chroma + BGE)")
    print(generate_answer(q, docs_c, llm_flan))

    print("\n🔹 Falcon (Chroma + BGE)")
    print(generate_answer(q, docs_c, llm_falcon))

    print("="*70)
```

```python
# ==============================
# MODEL EVALUATION + SUMMARY
# ==============================

queries = [
    "What is DFA?",
    "Explain NLP",
    "What is tokenization?"
]

results = {
    "GPT-2": [],
    "DistilGPT2": [],
    "FLAN-T5": [],
    "Falcon": []
}

# Run models
for q in queries:
    docs_f = db_faiss.similarity_search(q, k=3)
    docs_c = db_chroma.similarity_search(q, k=3)

    results["GPT-2"].append(generate_answer(q, docs_f, llm_gpt2))
    results["DistilGPT2"].append(generate_answer(q, docs_f, llm_distil))
    results["FLAN-T5"].append(generate_answer(q, docs_c, llm_flan))
    results["Falcon"].append(generate_answer(q, docs_c, llm_falcon))


# ==============================
# PRINT OUTPUTS
# ==============================

for i, q in enumerate(queries):
    print("\n" + "="*70)
    print("QUESTION:", q)

    for model in results:
        print(f"\n🔹 {model}:")
        print(results[model][i][:300])  # limit output for clean display


# ==============================
# SIMPLE QUALITY ANALYSIS
# ==============================

def evaluate_quality(text):
    length = len(text.split())

    if length < 10:
        return "Basic"
    elif length < 25:
        return "Slightly better"
    elif length < 60:
        return "Good but verbose"
    else:
        return "Best"


# ==============================
# FINAL SUMMARY TABLE
# ==============================

print("\n" + "="*70)
print("📊 FINAL MODEL COMPARISON")
print("="*70)

summary = {}

for model in results:
    scores = [evaluate_quality(ans) for ans in results[model]]

    # majority vote
    final = max(set(scores), key=scores.count)
    summary[model] = final

# Print table
print("\nModel\t\tOutput Quality")
print("-"*40)

for model, quality in summary.items():
    print(f"{model}\t\t{quality}")


# ==============================
# FOR REPORT (FINAL STATEMENT)
# ==============================

print("\nConclusion:")
print("FLAN-T5 performed best with accurate and structured answers.")
print("Falcon produced detailed but sometimes verbose responses.")
print("DistilGPT2 performed slightly better than GPT-2.")
print("GPT-2 produced basic outputs.")
```

```python
from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import (
    RecursiveCharacterTextSplitter,
    CharacterTextSplitter,
    TokenTextSplitter
)
from langchain.schema import Document
import re

# ==============================
# LOAD DOCUMENT
# ==============================
loader = PyPDFLoader("data.pdf")
documents = loader.load()

# ==============================
# 1. RECURSIVE
# ==============================
splitter1 = RecursiveCharacterTextSplitter(chunk_size=500, chunk_overlap=50)
chunks1 = splitter1.split_documents(documents)

# ==============================
# 2. CHARACTER
# ==============================
splitter_alt = CharacterTextSplitter(separator="\n", chunk_size=700, chunk_overlap=100)
chunks_alt = splitter_alt.split_documents(documents)

# ==============================
# 3. SEMANTIC
# ==============================
semantic_splitter = RecursiveCharacterTextSplitter(
    separators=["\n\n", "\n", ".", " "],
    chunk_size=800,
    chunk_overlap=150
)
chunks_semantic = semantic_splitter.split_documents(documents)

# ==============================
# 4. TOKEN
# ==============================
token_splitter = TokenTextSplitter(chunk_size=200, chunk_overlap=50)
chunks_token = token_splitter.split_documents(documents)

# ==============================
# 5. SENTENCE
# ==============================
def sentence_chunking(docs, chunk_size=3):
    chunks = []
    for doc in docs:
        sentences = re.split(r'(?<=[.!?]) +', doc.page_content)
        for i in range(0, len(sentences), chunk_size):
            chunk_text = " ".join(sentences[i:i+chunk_size])
            chunks.append(Document(page_content=chunk_text))
    return chunks

chunks_sentence = sentence_chunking(documents)

# ==============================
# 6. HYBRID
# ==============================
def hybrid_chunking(docs):
    chunks = []
    for doc in docs:
        words = doc.page_content.split()
        temp = []
        for word in words:
            temp.append(word)
            if len(temp) >= 150:
                chunks.append(Document(page_content=" ".join(temp)))
                temp = []
        if temp:
            chunks.append(Document(page_content=" ".join(temp)))
    return chunks

chunks_hybrid = hybrid_chunking(documents)

# ==============================
# 7. SLIDING WINDOW
# ==============================
def sliding_window_chunking(docs, window=100, step=50):
    chunks = []
    for doc in docs:
        words = doc.page_content.split()
        for i in range(0, len(words), step):
            chunk = " ".join(words[i:i+window])
            chunks.append(Document(page_content=chunk))
    return chunks

chunks_window = sliding_window_chunking(documents)

print("✅ All chunking methods created successfully")
```

```python
print("\n📊 CHUNKING COMPARISON")
print("="*50)

methods = {
    "Recursive": chunks1,
    "Character": chunks_alt,
    "Semantic": chunks_semantic,
    "Token": chunks_token,
    "Sentence": chunks_sentence,
    "Hybrid": chunks_hybrid,
    "Sliding Window": chunks_window
}

for name, ch in methods.items():
    print(f"{name}: {len(ch)} chunks")
```
